### Combinational Circuits


```python
from logic.combinational_circuits import half_adder, full_adder

C_S_half0 = half_adder(0, 0)
C_S_half1 = half_adder(0, 1)
C_S_half2 = half_adder(1, 0)
C_S_half3 = half_adder(1, 1)

print("HALF ADDER - TRUTH TABLE")
print(f"""
    x y |  C  S
    0 0 | {C_S_half0}
    0 1 | {C_S_half1}
    1 0 | {C_S_half2}
    1 1 | {C_S_half3}
    """)


C_S_full0 = full_adder(0, 0, 0)
C_S_full1 = full_adder(0, 0, 1)
C_S_full2 = full_adder(0, 1, 0)
C_S_full3 = full_adder(0, 1, 1)
C_S_full4 = full_adder(1, 0, 0)
C_S_full5 = full_adder(1, 0, 1)
C_S_full6 = full_adder(1, 1, 0)
C_S_full7 = full_adder(1, 1, 1)

print("FULL ADDER - TRUTH TABLE")
print(f"""
    x y z |  C  S
    0 0 0 | {C_S_full0}
    0 0 1 | {C_S_full1}
    0 1 0 | {C_S_full2}
    0 1 1 | {C_S_full3}
    1 0 0 | {C_S_full4}
    1 0 1 | {C_S_full5}
    1 1 0 | {C_S_full6}
    1 1 1 | {C_S_full7}
#     """)
```

```python
from logic.combinational_circuits import (
    binary_to_BCD_decoder,
    BCD_to_seven_segment_decoder,
)
from peripheral.output_devices import seven_segment_displays

b = binary(120, word=8)  # * Upto 255 (because highest 8-bit value is 255)

T1, T0, H3, H2, H1, H0, U3, U2, U1, U0 = binary_to_BCD_decoder(*b)

seven_segment_displays(
    BCD_to_seven_segment_decoder(0, 0, T1, T0),
    BCD_to_seven_segment_decoder(H3, H2, H1, H0),
    BCD_to_seven_segment_decoder(U3, U2, U1, U0),
)
```

## Flip-Flops


### SR Latch


```python
from logic.sequential_circuits import SRLatch
from clock.clocks import binary_clock

clk = binary_clock()

sr_latch = SRLatch()
print(sr_latch(0, next(clk)))
print(sr_latch(1, next(clk)))
print(sr_latch(1, next(clk)))
print(sr_latch(0, next(clk)))
print(sr_latch(1, next(clk)))
```

### ALU in Action


```python
from logic.logicgates import AND, OR, NOT
from logic.combinational_circuits import (
    binary_to_BCD_decoder,
    BCD_to_seven_segment_decoder,
)
from logic.alu import FourBitALU
from peripheral.output_devices import seven_segment_displays

alu = FourBitALU()
```

```python
print(alu)
```

```python
G1, G0 = 0, 0  # Group
F2, F1, F0 = 0, 1, 0  # Mnemonic

ACC = (0, 1, 1, 1)
B = (1, 0, 1, 1)

CIN = 0  # Carry in

#   ┌──────────┬─────────────────┬──────────┬─────────────────┬───────────────────┐
# * │ S = Sign │ CF = Carry Flag │ Z = Zero │ P = Parity Flag │ V = Overflow Flag │
#   └──────────┴─────────────────┴──────────┴─────────────────┴───────────────────┘
S, CF, Y, Z, P, V = alu(ACC, B, G1, G0, F2, F1, F0, CIN)

_, _, T3, T2, T1, T0, U3, U2, U1, U0 = binary_to_BCD_decoder(
    0,
    0,
    0,
    # * Checking wether the SUB (subtraction) operation is performing or not.
    # * SUB always produce a carry if the result is in positive.
    # * So we need to discard it when we perform SUB.
    (CF if NOT(OR(AND(F1, NOT(F0)), AND(NOT(F2), F1), AND(F2, NOT(F1), F0))) else 0),
    *Y,
)

# * Display Seven Segment Output in the terminal
seven_segment_displays(
    (0, 0, 0, 0, 0, 0, S),
    BCD_to_seven_segment_decoder(T3, T2, T1, T0),
    BCD_to_seven_segment_decoder(U3, U2, U1, U0),
)

print(f"{S = }, {CF = }, {Y = }")
print(f"{Z = }, {P = }, {V = }")
```
